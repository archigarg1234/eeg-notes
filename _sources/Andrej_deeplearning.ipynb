So in the backpropagation, what we are essentially looking for is, the change of the loss fuction if we chnage the weights by some value. thats what we also call a derivative of a function. 

```python
class Value:

    def __init__(self, data, _children = (), _op='', label = ''):
        self.data = data 
        self.grad = 0.0
        self._prev = set(_children)
        self._op = _op
        self.label = label
    def __repr__(self):
        return f"value(data={self.data})"
    def __add__(self, other):
        out = Value(self.data + other.data, (self, other), '+')
        return out
    def __mul__(self, other):
        out = Value(self.data * other.data, (self, other),'*')
        return out
    

a = Value(2.0, label = 'a')
b = Value(-3.0, label = 'b')
c = Value(10, label = 'c')
e = a*b ; e.label = 'e'
d = e +c ; d.label = 'd'
f= Value(-2.0, label= 'f')
L = d * f ; L.label = 'L'
L
#a+b
# a.__add__(b) # self == a , other == b

# (a.__mul__(b)).__add__(c)
```

## Chain rule of calculus 

We know, 
1- Change of L wrt to L. 
2- We can get the change of L wrt to d and f . derivative. 
3- Now if we have to get the change of L wrt to d or e, in that case, we have to apply the chain rule. 
Chain rule says : If you want to get the derivative of L wrt to d , then you need to multiply the derivative of L wrt to d and derivative of d wrt to c or e.

You can easily do this over pen and paper.

dL/ dc = (dL /dd )* (dd/dc)

so the gradient for c and e wil be same ad dL/dd because, "+ " just propagate the derivatives. You can easily do this with the pen and paper and its a simple task. 


So, the idea is, when we backpropaget, it tells us how does the loss function changes if I change the weights a bit. and when we find the gradient, it helps to guide the the change in the direction , where L becomes less.
