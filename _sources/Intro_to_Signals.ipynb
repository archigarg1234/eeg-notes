> **A signal is any quantity measured at regular intervals.**

Signals are everywhere — audio waveforms, stock prices, temperature readings, and of course, brain electrical activity (EEG). At its core, a signal is just a sequence of numbers that encode how something changes over time (or space).

### Continuous vs Discrete Signals

| Property | Continuous Signal | Discrete Signal |
|---|---|---|
| **Definition** | Defined at **every** instant in time | Defined only at **specific** time points |
| **Representation** | $s(t)$, where $t \in \mathbb{R}$ | $s[n]$, where $n \in \mathbb{Z}$ |
| **Example** | Voltage from an electrode | Digitized EEG after A/D conversion |

In practice, we almost always work with **discrete** signals — continuous real-world phenomena are **sampled** at regular time intervals by our recording equipment.

# 2. Key Terminology

```python
# Explore frequency and period
fs = 1000
freq = 3.0
t = np.arange(0, 1, 1/fs)
s = np.sin(2 * np.pi * freq * t)
T = 1 / freq

fig, ax = plt.subplots(figsize=(12, 4))
ax.plot(t, s, color='steelblue', linewidth=1.5)

# Highlight one period
ax.axvspan(0, T, alpha=0.15, color='orange', label=f'One period T = {T:.3f} s')
ax.axvline(T, color='orange', linestyle='--', alpha=0.7)

# Mark zero crossings
zero_crossings = [0, T/2, T, 3*T/2, 2*T]
for zc in zero_crossings:
    if zc <= 1:
        ax.plot(zc, 0, 'ro', markersize=6, zorder=5)

ax.set_title(f'Frequency = {freq:.1f} Hz  |  Period T = {T:.4f} s', fontsize=13)
ax.set_xlabel('Time (s)')
ax.set_ylabel('Amplitude')
ax.legend(fontsize=10)
ax.set_xlim(0, 1)
plt.tight_layout()
plt.show()
```

### Sampling Rate ($f_s$)

The **sampling rate** (or sampling frequency) is the number of samples captured per second, measured in **Hertz (Hz)**.

- A sampling rate of 256 Hz means 256 data points per second.
- It is also related to zero-crossings: the number of zero-crossings per second $\times$ 0.5 gives the dominant frequency.

### Frequency and Period

$$f = \frac{1}{T}$$

where:
- $f$ is the **frequency**: cycles per second (Hz),
- $T$ is the **period**: the duration of one complete cycle (seconds).

| Concept | Symbol | Unit | Meaning |
|---|---|---|---|
| Frequency | $f$ | Hz | How many cycles occur per second |
| Period | $T$ | seconds | How long one cycle takes |
| Sampling rate | $f_s$ | Hz | How many measurements per second |
| Sample interval | $\Delta t = 1/f_s$ | seconds | Time between consecutive samples |

# 3. Properties of Oscillations

```python
# Explore amplitude, frequency, and phase
fs = 1000
t = np.arange(0, 2, 1/fs)

amplitude = 3.0
frequency = 5.0
phase = np.pi/4

s = amplitude * np.sin(2 * np.pi * frequency * t + phase)
ref = np.sin(2 * np.pi * frequency * t)

fig, ax = plt.subplots(figsize=(12, 4))
ax.plot(t, s, color='steelblue', linewidth=1.5, label=f'a={amplitude}, f={frequency} Hz, phase={phase:.2f} rad')
ax.plot(t, ref, color='gray', alpha=0.3, linewidth=1, linestyle='--', label='Reference (a=1, phase=0)')

ax.axhline(0, color='black', linewidth=0.5)
ax.axhline(amplitude, color='red', linewidth=0.8, linestyle=':', alpha=0.6, label=f'Peak = {amplitude}')
ax.axhline(-amplitude, color='red', linewidth=0.8, linestyle=':', alpha=0.6)

ax.set_title(f'a = {amplitude:.1f}  |  f = {frequency:.1f} Hz  |  phase = {phase:.2f} rad ({np.degrees(phase):.0f} deg)', fontsize=13)
ax.set_xlabel('Time (s)')
ax.set_ylabel('Amplitude')
ax.legend(loc='upper right', fontsize=9)
plt.tight_layout()
plt.show()
```

Any oscillation is fully characterized by **three quantities**:

| Property | Question it answers | Unit |
|---|---|---|
| **Frequency** | How fast does it oscillate? | Hz |
| **Amplitude / Power** | How strong is it? | μV / μV² |
| **Phase** | When does the cycle start? (amplitude at $t = 0$) | radians |

Together, these three numbers can reconstruct any pure sinusoidal signal. This is the fundamental insight behind the **Fourier Transform** — any complex signal can be decomposed into a sum of sinusoids, each described by its frequency, amplitude, and phase.

# 4. Amplitude Measures

```python
# Demonstrate all three amplitude measures on a realistic signal
np.random.seed(42)
fs = 256
t = np.arange(0, 3, 1/fs)

# Composite signal: alpha + beta + noise
signal = (
    2.0 * np.sin(2 * np.pi * 10 * t)       # alpha
    + 0.8 * np.sin(2 * np.pi * 22 * t)     # beta
    + 0.5 * np.random.randn(len(t))         # noise
)

peak_amp = np.max(np.abs(signal))
peak_to_peak = np.max(signal) - np.min(signal)
rms_amp = np.sqrt(np.mean(signal**2))

fig, axes = plt.subplots(1, 3, figsize=(16, 4))

# Peak amplitude
axes[0].plot(t, signal, color='steelblue', linewidth=0.7)
idx_peak = np.argmax(np.abs(signal))
axes[0].axhline(peak_amp, color='red', linestyle='--', label=f'Peak = {peak_amp:.2f}')
axes[0].axhline(-peak_amp, color='red', linestyle='--')
axes[0].plot(t[idx_peak], signal[idx_peak], 'ro', markersize=8, zorder=5)
axes[0].set_title('Peak Amplitude', fontsize=12, fontweight='bold')
axes[0].legend(fontsize=9)

# Peak-to-peak
axes[1].plot(t, signal, color='steelblue', linewidth=0.7)
axes[1].axhline(np.max(signal), color='orange', linestyle='--')
axes[1].axhline(np.min(signal), color='orange', linestyle='--')
axes[1].annotate('', xy=(2.8, np.max(signal)), xytext=(2.8, np.min(signal)),
                 arrowprops=dict(arrowstyle='<->', color='orange', lw=2))
axes[1].text(2.85, 0, f'P2P = {peak_to_peak:.2f}', color='orange', fontsize=11, fontweight='bold')
axes[1].set_title('Peak-to-Peak Amplitude', fontsize=12, fontweight='bold')

# RMS
axes[2].plot(t, signal, color='steelblue', linewidth=0.7)
axes[2].axhline(rms_amp, color='green', linestyle='--', label=f'RMS = {rms_amp:.2f}')
axes[2].axhline(-rms_amp, color='green', linestyle='--')
axes[2].fill_between(t, -rms_amp, rms_amp, alpha=0.1, color='green')
axes[2].set_title('RMS Amplitude', fontsize=12, fontweight='bold')
axes[2].legend(fontsize=9)

for ax in axes:
    ax.set_xlabel('Time (s)')
    ax.set_ylabel('Amplitude (uV)')

plt.tight_layout()
plt.show()

print(f'Peak Amplitude:      {peak_amp:.3f} uV')
print(f'Peak-to-Peak:        {peak_to_peak:.3f} uV')
print(f'RMS Amplitude:       {rms_amp:.3f} uV')
```

There are several ways to quantify the "strength" of a signal:

### Peak Amplitude
The maximum absolute value of the signal. Measures the strength of **individual EEG components** (e.g., amplitude of an ERP or spike). Useful for detecting **abnormal spikes** (e.g., epileptic discharges).

### Peak-to-Peak Amplitude
The difference between the maximum and minimum values. Reflects the **overall variability** of the EEG signal. Used to assess **signal-to-noise ratio** or amplitude modulation.

$$A_{p2p} = \max(s) - \min(s)$$

### RMS Amplitude (Root Mean Square)
The square root of the average of the squared values — the **effective amplitude** accounting for both positive and negative fluctuations:

$$A_{\text{RMS}} = \sqrt{\frac{1}{N} \sum_{i=1}^{N} s_i^2}$$

| Measure | What it captures | Best for |
|---|---|---|
| Peak | Maximum excursion | Spike detection, ERPs |
| Peak-to-peak | Total range of variation | SNR, oscillation strength |
| RMS | Effective average magnitude | Overall signal power, comparing conditions |

```python
# Sliding window RMS: see how amplitude varies over time
np.random.seed(42)
fs = 256
t = np.arange(0, 4, 1/fs)

# Signal with amplitude modulation (mimics EEG alpha bursting)
envelope = 1 + 1.5 * np.exp(-((t - 2.0)**2) / (2 * 0.3**2))
signal = envelope * np.sin(2 * np.pi * 10 * t) + 0.3 * np.random.randn(len(t))

# Compute sliding-window RMS (200 ms window)
window_ms = 200
win_samples = int(window_ms / 1000 * fs)
rms = np.array([
    np.sqrt(np.mean(signal[max(0, i - win_samples//2):i + win_samples//2]**2))
    for i in range(len(signal))
])

fig, ax = plt.subplots(figsize=(12, 4))
ax.plot(t, signal, color='steelblue', alpha=0.5, linewidth=0.6, label='Signal')
ax.plot(t, rms, color='red', linewidth=2, label=f'Sliding RMS ({window_ms} ms window)')
ax.plot(t, -rms, color='red', linewidth=2)
ax.set_xlabel('Time (s)')
ax.set_ylabel('Amplitude (uV)')
ax.set_title('Sliding-Window RMS Amplitude', fontsize=13)
ax.legend(fontsize=10)
plt.tight_layout()
plt.show()
```

# 5. Noise

> **Noise is any unwanted signal component that contaminates the measurement.**

```python
# Generate and compare noise types
np.random.seed(0)
fs = 1000
duration = 10
t = np.arange(0, duration, 1/fs)
n = len(t)

# Gaussian white noise
gwn = np.random.randn(n)

# Uniform white noise (shifted to zero-mean)
uwn = np.random.rand(n) - 0.5

# Pink noise (1/f): shape white noise spectrum
white_fft = np.fft.rfft(np.random.randn(n))
freqs = np.fft.rfftfreq(n, d=1/fs)
freqs[0] = 1  # avoid division by zero
pink_fft = white_fft / np.sqrt(freqs)
pink = np.fft.irfft(pink_fft, n=n)
pink /= np.std(pink)  # normalize

noises = [('Gaussian White Noise', gwn, 'steelblue'),
          ('Uniform White Noise', uwn, 'teal'),
          ('Pink Noise (1/f)', pink, 'indianred')]

fig, axes = plt.subplots(3, 1, figsize=(14, 8), sharex=True)
for ax, (title, noise, color) in zip(axes, noises):
    ax.plot(t[:2000], noise[:2000], color=color, linewidth=0.5, alpha=0.8)
    ax.set_title(title, fontsize=12, fontweight='bold')
    ax.set_ylabel('Amplitude')
axes[-1].set_xlabel('Time (s)')
plt.tight_layout()
plt.show()
```

Understanding noise is critical because **all real signals are noisy**.

### White Noise
**Equal power at all frequencies** — the spectrum is flat. It can be:
- **Gaussian (normally distributed):** Amplitudes follow a bell curve $\mathcal{N}(0, \sigma^2)$. Most common model for electronic/thermal noise.
- **Uniform:** Amplitudes are equally likely across a range.

### Pink Noise (1/f Noise)
Power **decreases** with increasing frequency, following a $1/f$ characteristic:
- Low frequencies are loud, high frequencies are quiet.
- **Very common in biological signals**, including EEG — the brain's background activity has a natural $1/f$ spectrum.

| Noise Type | Spectrum Shape | Amplitude Distribution | Where Found |
|---|---|---|---|
| Gaussian White | Flat | Normal (bell curve) | Electronics, thermal |
| Uniform White | Flat | Uniform (rectangular) | Quantization, simulation |
| Pink (1/f) | $\propto 1/f$ | Varies | Biological signals, EEG |

```python
# Power Spectral Density comparison
from scipy.signal import welch

fig, axes = plt.subplots(1, 3, figsize=(16, 4))

for ax, (title, noise, color) in zip(axes, noises):
    f_psd, psd = welch(noise, fs=fs, nperseg=1024)
    ax.semilogy(f_psd, psd, color=color, linewidth=1)
    ax.set_title(f'PSD: {title}', fontsize=11, fontweight='bold')
    ax.set_xlabel('Frequency (Hz)')
    ax.set_ylabel('Power/Hz (log scale)')
    ax.set_xlim(0, 100)

plt.tight_layout()
plt.show()
```

```python
# Amplitude distributions
fig, axes = plt.subplots(1, 3, figsize=(16, 4))

for ax, (title, noise, color) in zip(axes, noises):
    ax.hist(noise, bins=60, density=True, color=color, alpha=0.7, edgecolor='white')
    ax.axvline(np.median(noise), color='black', linestyle='--', linewidth=1.5,
               label=f'Median = {np.median(noise):.3f}')
    ax.set_title(f'Distribution: {title}', fontsize=11, fontweight='bold')
    ax.set_xlabel('Amplitude')
    ax.set_ylabel('Density')
    ax.legend(fontsize=9)

plt.tight_layout()
plt.show()
```

### 6. The Nyquist Theorem — *Why* 2x?

The Nyquist theorem states:

$$f_{\max} = \frac{f_s}{2}$$

You can only faithfully capture frequencies **up to half** your sampling rate.

Suppose we sample at $f_s = 100$ Hz — that's one sample every $\frac{1}{100} = 0.01$ seconds.

Now consider a **50 Hz signal** — it completes one full cycle in $\frac{1}{50} = 0.02$ seconds. At 100 Hz sampling, we get **exactly 2 samples per cycle** — one at the peak, one at the trough. That's the bare minimum needed to see "something is oscillating." 50 Hz is the **Nyquist frequency** — the absolute ceiling.

What happens if we try to capture **90 Hz** with our 100 Hz sampling? A 90 Hz wave completes one cycle in $\frac{1}{90} \approx 0.011$ seconds, but our samples are spaced 0.01 seconds apart. We don't have enough samples to track it — instead, the samples *happen to line up* with a slow 10 Hz wave. The 90 Hz signal **mistaken as 10 Hz**. This is **aliasing**:

$$f_{\text{alias}} = f_s - f_{\text{true}} = 100 - 90 = 10 \text{ Hz}$$

Any frequency above Nyquist ($f_s/2$) gets "folded back" into the range $[0, f_s/2]$, producing **false information** that's indistinguishable from real low-frequency activity.


```python
# ── Nyquist Theorem & Aliasing ───────────────────────────────────────────
import numpy as np
import matplotlib.pyplot as plt

fs = 100                               # sampling rate (Hz)
t_cont = np.arange(0, 0.2, 0.00005)   # "continuous" time
t_samp = np.arange(0, 0.2, 1/fs)      # sample times

fig, axes = plt.subplots(3, 1, figsize=(14, 10))

# ── Panel 1: 10 Hz — easy, well sampled ──────────────────────────────────
freq = 10
s_true = np.sin(2 * np.pi * freq * t_cont)
s_samp = np.sin(2 * np.pi * freq * t_samp)

axes[0].plot(t_cont, s_true, color='steelblue', lw=1.5, label=f'True signal ({freq} Hz)')
axes[0].stem(t_samp, s_samp, linefmt='teal', markerfmt='o', basefmt=' ',
             label=f'Samples (fs={fs} Hz)')
axes[0].set_title(f'{freq} Hz — 10 samples per cycle  |  Well above Nyquist  ✓',
                  fontsize=13, fontweight='bold', color='#2e6e2e')
axes[0].set_ylabel('Amplitude')
axes[0].legend(loc='upper right', fontsize=9)

# ── Panel 2: 50 Hz — Nyquist limit, barely captured ──────────────────────
freq = 50
s_true = np.sin(2 * np.pi * freq * t_cont)
s_samp = np.sin(2 * np.pi * freq * t_samp)

axes[1].plot(t_cont, s_true, color='steelblue', lw=1.5, label=f'True signal ({freq} Hz)')
axes[1].stem(t_samp, s_samp, linefmt='teal', markerfmt='o', basefmt=' ',
             label=f'Samples (fs={fs} Hz)')

# Annotate one cycle
T = 1/freq
axes[1].annotate('', xy=(0, -1.3), xytext=(T, -1.3),
                 arrowprops=dict(arrowstyle='<->', color='orange', lw=2))
axes[1].text(T/2, -1.5, f'1 cycle = {T*1000:.0f} ms\n2 samples per cycle',
             ha='center', color='orange', fontsize=10, fontweight='bold')

axes[1].set_title(f'{freq} Hz — Nyquist limit (fs/2 = {fs//2} Hz)  |  '
                  f'Exactly 2 samples/cycle — barely captured  ⚠',
                  fontsize=13, fontweight='bold', color='#b8860b')
axes[1].set_ylabel('Amplitude')
axes[1].legend(loc='upper right', fontsize=9)

# ── Panel 3: 90 Hz — aliased as 10 Hz ────────────────────────────────────
freq = 90
alias_freq = fs - freq   # = 10 Hz
s_true  = np.sin(2 * np.pi * freq * t_cont)
s_samp  = np.sin(2 * np.pi * freq * t_samp)
s_alias = np.sin(2 * np.pi * alias_freq * t_cont)

axes[2].plot(t_cont, s_true, color='#e74c3c', lw=1.2, alpha=0.4,
             label=f'True signal ({freq} Hz)')
axes[2].plot(t_cont, s_alias, color='#2ecc71', lw=2.5, ls='--',
             label=f'Alias ({alias_freq} Hz) — what we reconstruct')
axes[2].stem(t_samp, s_samp, linefmt='teal', markerfmt='o', basefmt=' ',
             label=f'Samples (fs={fs} Hz)')

# Arrow showing the fold
axes[2].annotate(
    f'90 Hz folds to\n{fs} − 90 = {alias_freq} Hz',
    xy=(0.07, s_samp[7] if len(s_samp) > 7 else 0),
    xytext=(0.12, 1.2),
    fontsize=11, fontweight='bold', color='#e74c3c',
    arrowprops=dict(arrowstyle='->', color='#e74c3c', lw=1.5),
)

axes[2].set_title(f'{freq} Hz — Above Nyquist  |  Aliased as {alias_freq} Hz  ✗  FALSE INFORMATION',
                  fontsize=13, fontweight='bold', color='#b22222')
axes[2].set_ylabel('Amplitude')
axes[2].set_xlabel('Time (s)')
axes[2].legend(loc='upper right', fontsize=9)

for ax in axes:
    ax.set_xlim(0, 0.2)
    ax.axhline(0, color='#cccccc', lw=0.5)
    ax.grid(True, alpha=0.2)

plt.tight_layout()
plt.show()
```

# 7. The Sine Wave

The sine wave is the fundamental building block of signal processing. Every signal can be decomposed into a sum of sine waves (the core idea of the Fourier Transform).

### General Formula

$$y = a \times \sin(2\pi \times f \times t + \theta)$$

where:
- $a$ is the **amplitude** (peak height),
- $f$ is the **frequency** in Hz (cycles per second),
- $t$ is **time** in seconds,
- $\theta$ is the **phase angle** in radians.

### Sine vs Cosine

A cosine is just a **phase-shifted sine**:
- $\theta = 0$: **sine** — starts at zero, rises first
- $\theta = \pi/2$: **cosine** — starts at the peak

$$\cos(t) = \sin\left(t + \frac{\pi}{2}\right)$$

```python
# Superposition: adding multiple sine waves
fs = 256
t = np.arange(0, 2 * np.pi, 1/fs)

amplitudes  = [10, 5, 2]
frequencies = [2, 5, 10]
phases      = [0, np.pi, np.pi/2]
colors      = ['#e74c3c', '#2ecc71', '#9b59b6']

fig, axes = plt.subplots(len(amplitudes) + 1, 1, figsize=(12, 10), sharex=True)

composite = np.zeros(len(t))
for i, (a, f, p, c) in enumerate(zip(amplitudes, frequencies, phases, colors)):
    component = a * np.sin(2 * np.pi * f * t + p)
    composite += component
    axes[i].plot(t, component, color=c, linewidth=1.2)
    axes[i].set_title(f'Component {i+1}: a={a}, f={f} Hz, phase={p:.2f} rad', fontsize=11)
    axes[i].set_ylabel('Amplitude')

axes[-1].plot(t, composite, color='steelblue', linewidth=1.5)
axes[-1].set_title('Composite Signal (sum of all components)', fontsize=12, fontweight='bold')
axes[-1].set_xlabel('Time (s)')
axes[-1].set_ylabel('Amplitude')

fig.suptitle('Superposition Principle: Complex Signals = Sum of Simple Sinusoids', fontsize=14, y=1.01)
plt.tight_layout()
plt.show()
```

# 8. Euler's Formula & Complex Exponentials

We started by understanding oscillations using sine waves. A sine wave is fully described by three things: **amplitude**, **frequency**, and **phase**.

Now, instead of viewing oscillations only as up-and-down waves, we can think of them **geometrically** — as a point rotating around a circle.

A point rotating in a circle has:
- a horizontal projection → **cosine**
- a vertical projection → **sine**

This circular motion can be compactly represented using **complex numbers**. Euler's formula connects these two ideas:

$$e^{i\theta} = \cos(\theta) + i\sin(\theta)$$

A complex exponential **is** a rotating vector, where:
- the **real part** is cosine
- the **imaginary part** is sine

If the angle changes with time:

$$e^{i \cdot 2\pi f t}$$

then the vector **rotates at frequency $f$**.

---

### Why does this matter? → The Fourier Transform

This idea becomes the foundation of the **Fourier Transform**.

The Fourier Transform asks:

> *"How much of each rotating frequency component exists inside the signal?"*

To answer this, the signal is **compared** (multiplied and summed) with rotating complex exponentials at different frequencies.

- If the signal **contains** a particular frequency → the alignment is strong → **large value**
- If the frequency is **absent** → the alignment cancels out over time → **near zero**

So the Fourier Transform essentially measures **how strongly the signal matches oscillations at different frequencies**. This is how any complex signal can be decomposed into its underlying frequency components.


# 9. The Gaussian Function

```python
# Gaussian with different parameters
x = np.linspace(-10, 10, 1000)

fig, ax = plt.subplots(figsize=(10, 4))
for a, b, c, label, color in [(1, 0, 1, 'a=1, b=0, c=1', 'steelblue'),
                                (2, 0, 1, 'a=2, b=0, c=1', '#e74c3c'),
                                (1, 2, 0.5, 'a=1, b=2, c=0.5', '#2ecc71')]:
    g = a * np.exp(-((x - b)**2) / (2 * c**2))
    ax.plot(x, g, linewidth=2, label=label, color=color)

ax.set_title('Gaussian Function with different parameters', fontsize=13)
ax.set_xlabel('x')
ax.set_ylabel('f(x)')
ax.legend(fontsize=10)
ax.set_xlim(-5, 5)
plt.tight_layout()
plt.show()
```

The Gaussian function is a symmetric, **bell-shaped curve** that appears everywhere in signal processing and statistics.

### Mathematical Formulation

$$f(x) = a \times e^{-\frac{(x - b)^2}{2c^2}}$$

where:
- $a$ controls the **peak height**,
- $b$ determines the **position** of the peak (the center/mean),
- $c$ controls the **width** (related to the standard deviation).

### Why is the Gaussian Special?

1. **Self-similar under Fourier Transform:** A Gaussian in time is also a Gaussian in frequency — the optimal trade-off between time and frequency localization.
2. **Central Limit Theorem:** The sum of many independent random variables converges to a Gaussian.

### Applications in EEG Analysis

| Application | How Gaussians are used |
|---|---|
| **Wavelet Transform** | Gaussian-derived wavelets (e.g., **Morlet wavelet**) balance time and frequency resolution |
| **Smoothing** | Gaussian kernels smooth EEG signals, reducing noise while preserving features |
| **Gaussian Filtering** | Removes high-frequency noise by weighting nearby samples with a Gaussian kernel |
| **Wavelet Analysis (CWT)** | A Gaussian-modulated complex sinusoid (Morlet wavelet) decomposes EEG into time-frequency components |

```python
# Morlet wavelet: a Gaussian-windowed complex sinusoid
t = np.linspace(-3, 3, 1000)
freq = 2
sigma = 0.8

gaussian_env = np.exp(-t**2 / (2 * sigma**2))
sine_wave = np.cos(2 * np.pi * freq * t)
morlet = gaussian_env * sine_wave
morlet_imag = gaussian_env * np.sin(2 * np.pi * freq * t)

fig, axes = plt.subplots(2, 2, figsize=(14, 7))

axes[0, 0].plot(t, gaussian_env, color='orange', linewidth=2)
axes[0, 0].fill_between(t, gaussian_env, alpha=0.1, color='orange')
axes[0, 0].set_title('Gaussian Envelope', fontsize=12, fontweight='bold')

axes[0, 1].plot(t, sine_wave, color='steelblue', linewidth=1.5)
axes[0, 1].set_title(f'Cosine Wave ({freq} Hz)', fontsize=12, fontweight='bold')

axes[1, 0].plot(t, morlet, color='teal', linewidth=2)
axes[1, 0].plot(t, gaussian_env, color='orange', linewidth=1, linestyle='--', alpha=0.6)
axes[1, 0].plot(t, -gaussian_env, color='orange', linewidth=1, linestyle='--', alpha=0.6)
axes[1, 0].set_title('Morlet Wavelet (real) = Gaussian x Cosine', fontsize=12, fontweight='bold')

axes[1, 1].plot(t, morlet_imag, color='#9b59b6', linewidth=2)
axes[1, 1].plot(t, gaussian_env, color='orange', linewidth=1, linestyle='--', alpha=0.6)
axes[1, 1].plot(t, -gaussian_env, color='orange', linewidth=1, linestyle='--', alpha=0.6)
axes[1, 1].set_title('Morlet Wavelet (imag) = Gaussian x Sine', fontsize=12, fontweight='bold')

for ax in axes.flat:
    ax.set_xlabel('Time (s)')
    ax.set_ylabel('Amplitude')

fig.suptitle('Building the Morlet Wavelet -- the workhorse of EEG time-frequency analysis',
             fontsize=14, y=1.02)
plt.tight_layout()
plt.show()
```

# 10. Energy and Power

### Why do we need "power"?

Two signals can have the **same frequency** but completely different **strength**. A quiet alpha hum and a roaring alpha burst are both 10 Hz — frequency alone can't tell them apart. 

>**Power** is the quantity that captures *how much* activity a signal carries.


```python
# ── Weak vs Strong alpha: same frequency, different power ─────────────────
import numpy as np
import plotly.graph_objects as go
from plotly.subplots import make_subplots
from IPython.display import HTML

fs = 256
t = np.arange(0, 2, 1/fs)
weak   = 2  * np.sin(2 * np.pi * 10 * t)
strong = 10 * np.sin(2 * np.pi * 10 * t)

# Animate: signal draws itself left to right
n_frames = 60
indices = np.linspace(10, len(t), n_frames, dtype=int)

fig = make_subplots(rows=1, cols=2, subplot_titles=(
    'Weak Alpha (2 µV)', 'Strong Alpha (10 µV)'))

fig.add_trace(go.Scatter(x=[], y=[], mode='lines',
              line=dict(color='#3498db', width=2), showlegend=False), row=1, col=1)
fig.add_trace(go.Scatter(x=[], y=[], mode='lines',
              line=dict(color='#e74c3c', width=2), showlegend=False), row=1, col=2)

frames = []
for idx in indices:
    frames.append(go.Frame(data=[
        go.Scatter(x=t[:idx], y=weak[:idx], mode='lines',
                   line=dict(color='#3498db', width=2)),
        go.Scatter(x=t[:idx], y=strong[:idx], mode='lines',
                   line=dict(color='#e74c3c', width=2)),
    ]))
fig.frames = frames

fig.update_xaxes(title_text='Time (s)', range=[0, 2])
fig.update_yaxes(range=[-12, 12])
fig.update_layout(height=350, width=750,
    title=dict(text='Same Frequency (10 Hz) — Different Power', font=dict(size=14)),
    annotations=[
        dict(text='Both are 10 Hz alpha — but one carries <b>25x more power</b>',
             xref='paper', yref='paper', x=0.5, y=-0.15,
             showarrow=False, font=dict(size=12, color='gray'))
    ])

html = fig.to_html(include_plotlyjs='cdn', full_html=False)
html += """
<script>
(function() {
    var el = document.currentScript.previousElementSibling;
    function loop() {
        Plotly.animate(el, null, {
            frame: {duration: 40, redraw: true},
            transition: {duration: 0},
            mode: 'immediate', fromcurrent: false
        }).then(loop);
    }
    setTimeout(loop, 500);
})();
</script>
"""
display(HTML(html))
```

Power is calculated as the **square the signal**

$$P(t) = s^2(t)$$

Squaring does three things:
- Removes negative signs (everything becomes positive)
- Makes large amplitudes contribute **much** more than small ones
- Gives a value proportional to signal strength

### From Time Domain to Frequency Domain: Band Power

When we apply the FFT, the total power gets **distributed across frequencies**. This lets us ask targeted questions:

- *How much power exists around 10 Hz?* → **Alpha power**
- *Is beta activity increasing during a task?* → **Beta power** (13–30 Hz)
- *Is there high-frequency gamma during memory encoding?* → **Gamma power** (30–45 Hz)

These are the **band powers** that EEG researchers talk about constantly.

| Band | Frequency | Associated with |
|---|---|---|
| Delta | 0.5 – 4 Hz | Deep sleep |
| Theta | 4 – 8 Hz | Memory, drowsiness |
| Alpha | 8 – 13 Hz | Relaxation, eyes closed |
| Beta | 13 – 30 Hz | Active thinking, focus |
| Gamma | 30 – 45 Hz | Perception, binding |


```python
# ── Band power: frequency domain animation ────────────────────────────────
import numpy as np
import plotly.graph_objects as go
from plotly.subplots import make_subplots
from IPython.display import HTML
import warnings
warnings.filterwarnings('ignore')


np.random.seed(42)
fs = 256
t = np.arange(0, 4, 1/fs)

signal = (3.0 * np.sin(2 * np.pi * 10 * t)        # alpha
          + 1.5 * np.sin(2 * np.pi * 22 * t)       # beta
          + 0.8 * np.sin(2 * np.pi * 6 * t)        # theta
          + 0.5 * np.random.randn(len(t)))          # noise

N = len(signal)
freqs = np.fft.rfftfreq(N, 1/fs)
power = (2.0 / N) * np.abs(np.fft.rfft(signal))**2

# Band definitions
bands = [
    ('Delta', 0.5, 4, 'rgba(128,128,128,0.3)', 'gray'),
    ('Theta', 4, 8, 'rgba(155,89,182,0.3)', '#9b59b6'),
    ('Alpha', 8, 13, 'rgba(46,204,113,0.3)', '#2ecc71'),
    ('Beta', 13, 30, 'rgba(52,152,219,0.3)', '#3498db'),
    ('Gamma', 30, 45, 'rgba(231,76,60,0.3)', '#e74c3c'),
]

# Animate: highlight one band at a time, cycling
fig = go.Figure()

# Base power spectrum (always visible)
fig.add_trace(go.Scatter(x=freqs, y=power, mode='lines',
    line=dict(color='steelblue', width=1.5), fill='tozeroy',
    fillcolor='rgba(70,130,180,0.1)', showlegend=False))

frames = []
for band_name, flo, fhi, fillcolor, linecolor in bands:
    band_mask = (freqs >= flo) & (freqs <= fhi)
    band_power = np.trapz(power[band_mask], freqs[band_mask])
    total_power = np.trapz(power, freqs)
    pct = band_power / total_power * 100

    highlight_y = np.where(band_mask, power, 0)

    frames.append(go.Frame(data=[
        go.Scatter(x=freqs, y=power, mode='lines',
                   line=dict(color='steelblue', width=1.5), fill='tozeroy',
                   fillcolor='rgba(70,130,180,0.1)', showlegend=False),
        go.Scatter(x=freqs, y=highlight_y, mode='lines', fill='tozeroy',
                   line=dict(color=linecolor, width=2.5),
                   fillcolor=fillcolor, showlegend=False),
    ],
    layout=go.Layout(
        title=dict(text=f'Power Spectrum — <b>{band_name}</b> band ({flo}–{fhi} Hz) = <b>{pct:.1f}%</b> of total power',
                   font=dict(size=14)),
    )))

# Repeat each band for ~1.5 seconds (30 frames at 50ms each)
looped_frames = []
for f in frames:
    looped_frames.extend([f] * 30)
fig.frames = looped_frames

fig.update_xaxes(title_text='Frequency (Hz)', range=[0, 50])
fig.update_yaxes(title_text='Power (µV²/Hz)')
fig.update_layout(height=400, width=750,
    title=dict(text='Power Spectrum — Band Power', font=dict(size=14)))

html = fig.to_html(include_plotlyjs=False, full_html=False)
html += """
<script>
(function() {
    var el = document.currentScript.previousElementSibling;
    function loop() {
        Plotly.animate(el, null, {
            frame: {duration: 50, redraw: true},
            transition: {duration: 0},
            mode: 'immediate', fromcurrent: false
        }).then(loop);
    }
    setTimeout(loop, 500);
})();
</script>
"""
display(HTML(html))
```

### Parseval's Theorem

Here's the key insight that ties everything together:

> **The total energy of a signal is the same whether you compute it in the time domain or the frequency domain.**

$$\sum_{n} |s[n]|^2 = \frac{1}{N} \sum_{k} |S[k]|^2$$

This means the power spectrum isn't just a visualization trick — it's a **mathematically equivalent** representation of the signal's energy. When you compute "alpha power" from the FFT, you're measuring *real physical energy* in that frequency range.


```python
# Parseval's theorem: energy in time domain = energy in frequency domain
np.random.seed(42)
fs = 256
t = np.arange(0, 4, 1/fs)
signal = (
    2 * np.sin(2 * np.pi * 10 * t)
    + 1 * np.sin(2 * np.pi * 22 * t)
    + 0.5 * np.random.randn(len(t))
)

N = len(signal)
energy_time = np.sum(signal**2)
fft_vals = np.fft.fft(signal)
energy_freq = np.sum(np.abs(fft_vals)**2) / N

freqs = np.fft.rfftfreq(N, 1/fs)
power_spectrum = (2.0 / N) * np.abs(np.fft.rfft(signal))**2

fig, axes = plt.subplots(1, 2, figsize=(14, 4))

axes[0].plot(t, signal, color='steelblue', linewidth=0.6)
axes[0].set_title(f'Time Domain  |  Energy = {energy_time:.1f}', fontsize=12)
axes[0].set_xlabel('Time (s)')
axes[0].set_ylabel('Amplitude (uV)')

axes[1].plot(freqs, power_spectrum, color='orange', linewidth=1.2)
axes[1].set_title(f'Power Spectrum  |  Energy = {energy_freq:.1f}', fontsize=12)
axes[1].set_xlabel('Frequency (Hz)')
axes[1].set_ylabel('Power (uV^2/Hz)')
axes[1].set_xlim(0, 50)

fig.suptitle("Parseval's Theorem: Energy is conserved across domains", fontsize=13, y=1.02)
plt.tight_layout()
plt.show()

print(f'Time-domain energy:      {energy_time:.2f}')
print(f'Frequency-domain energy: {energy_freq:.2f}')
print(f'Match: {np.isclose(energy_time, energy_freq, rtol=1e-10)}')
```
